```python
import pandas as pd
import math
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv(r"C:\AIml_intern\data\hiring.csv")
df
```

```python
df.experience = df.experience.fillna(0)
```

```python
from word2number import w2n

df.experience = df.experience.apply(lambda x: w2n.word_to_num(x) if type(x)==str else x)
```

```python
df.experience = df.experience.replace('two',2)
```

```python
df.experience = df.experience.fillna(0)
```

```python
median_test = math.floor(df['test_score(out of 10)'].median())
df['test_score(out of 10)'] = df['test_score(out of 10)'].fillna(median_test)
```

```python
X = df[['experience','test_score(out of 10)','interview_score(out of 10)']]
y = df['salary($)']
```

```python
model = LinearRegression()
model.fit(X,y)
```

```python
model.predict(pd.DataFrame([[2,9,6]], columns=[
    'experience',
    'test_score(out of 10)',
    'interview_score(out of 10)'
]))
```

```python
model.predict(pd.DataFrame([[12,10,10]], columns=[
    'experience',
    'test_score(out of 10)',
    'interview_score(out of 10)'
]))
```
