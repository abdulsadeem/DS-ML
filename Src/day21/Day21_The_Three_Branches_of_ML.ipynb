## The Three Branches of Machine Learning

### 1. Supervised Learning
- **Definition:** Learning from labeled data; the model knows the correct answers during training.
- **Goal:** Predict outcomes based on input features.
- **Examples:**
  - Spam detection (Emails labeled as spam or not)
  - House price prediction (Data labeled with prices)

### 2. Unsupervised Learning
- **Definition:** Learning from unlabeled data; the model tries to find hidden patterns or clusters.
- **Goal:** Discover structure in data.
- **Examples:**
  - Customer segmentation (Grouping customers based on behavior)
  - Market basket analysis (Finding products often bought together)

### 3. Reinforcement Learning
- **Definition:** Learning by interacting with an environment; receives rewards or penalties based on actions.
- **Goal:** Learn a strategy (policy) to maximize cumulative reward.
- **Examples:**
  - Self-driving cars (Reward for safe driving, penalty for collisions)
  - Game AI (Chess, Go, video games)

### Comparison Table

| ML Type | Labeled Data? | Goal | Real-World Examples |
|---------|---------------|------|-------------------|
| Supervised Learning | ✅ Yes | Predict outcomes | Spam detection, House price prediction |
| Unsupervised Learning | ❌ No | Find hidden patterns | Customer segmentation, Market basket analysis |
| Reinforcement Learning | N/A (via interaction) | Learn optimal strategy | Self-driving cars, Game AI |
